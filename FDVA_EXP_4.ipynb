```python
# Import required library
import pandas as pd
import numpy as np


# --------------------------------------------------
# 1. CREATE SAMPLE LOAN DATASET
# --------------------------------------------------

data = {
    'Gender': ['Male', 'Female', 'Male', 'Male', 'Female', 'Male', 'Female'],
    'Married': ['Yes', 'No', 'Yes', 'Yes', 'No', 'Yes', 'No'],
    'Dependents': ['0', '1', '3+', '2', '0', '3+', '1'],
    'Education': ['Graduate', 'Graduate', 'Not Graduate',
                  'Graduate', 'Graduate', 'Not Graduate', 'Graduate'],
    'Self_Employed': ['No', 'Yes', 'No', 'No', 'Yes', 'No', 'Yes'],
    'Property_Area': ['Urban', 'Rural', 'Semiurban',
                      'Urban', 'Rural', 'Semiurban', 'Urban'],
    'ApplicantIncome': [5000, 3000, 7000, 4500, 3500, 8000, 4000],
    'CoapplicantIncome': [1500, 0, 2000, 1000, 1200, 2500, 500],
    'LoanAmount': [200, 150, 300, 180, 160, 350, 175]
}

df = pd.DataFrame(data)


# --------------------------------------------------
# 2. DISPLAY FIRST AND LAST ROWS
# --------------------------------------------------

print("First 5 rows:")
print(df.head())

print("\nLast 5 rows:")
print(df.tail())


# --------------------------------------------------
# 3. BASIC INFORMATION
# --------------------------------------------------

print("\nDataset Information:")
df.info()


# --------------------------------------------------
# 4. DESCRIPTIVE STATISTICS
# --------------------------------------------------

print("\nDescriptive Statistics:")
print(df.describe())


# --------------------------------------------------
# 5. DISPLAY COLUMN NAMES
# --------------------------------------------------

print("\nColumn Names:")
print(df.columns)


# --------------------------------------------------
# 6. UNIQUE VALUES OF CATEGORICAL COLUMNS
# --------------------------------------------------

categorical_cols = [
    'Gender',
    'Married',
    'Dependents',
    'Education',
    'Self_Employed',
    'Property_Area'
]

for col in categorical_cols:
    unique_vals = df[col].unique()

    print(f"\nUnique values in '{col}':")
    print(unique_vals)


# --------------------------------------------------
# 7. CONVERT DEPENDENTS INTO NUMERIC
# --------------------------------------------------

# Replace '3+' with 3
df['Dependents'] = df['Dependents'].replace('3+', 3)

# Convert to numeric
df['Dependents'] = pd.to_numeric(
    df['Dependents'],
    errors='coerce'
).astype('Int64')


# Display value counts
print("\nValue counts in 'Dependents' after conversion:")
print(df['Dependents'].value_counts(dropna=False))


# --------------------------------------------------
# 8. STATISTICAL ANALYSIS
# --------------------------------------------------

cols = [
    'ApplicantIncome',
    'CoapplicantIncome',
    'LoanAmount'
]

for col in cols:

    print(f"\nStatistics for '{col}':")

    print(f"Mean: {df[col].mean():.2f}")

    print(f"Median: {df[col].median():.2f}")

    print(f"Mode: {df[col].mode().values}")

    print(f"Range: {df[col].max() - df[col].min():.2f}")

    print(f"Variance: {df[col].var():.2f}")

    print(f"Standard Deviation: {df[col].std():.2f}")
```
